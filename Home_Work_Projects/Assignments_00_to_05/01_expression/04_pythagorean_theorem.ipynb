```python
import math

def hypotenuse():
  ab : float = float(input("Enter the length of side AB: "))
  ac : float = float(input("Enter the length of side BC: "))
  bc : float = math.sqrt(ab**2 + ac**2)
  print("\nThe length of BC (the hypotenuse) is: " , str(bc))

hypotenuse()
```
