```python
def divide_numbers():
  dividend = int(input("Please enter an integar to be divided: "))
  divisor = int(input("Please enter an integar to divide by: "))
  quotient = dividend // divisor
  remainder = dividend % divisor
  print(f"\n{dividend} divided by {divisor} is {quotient} with a remainder of {remainder}.")

divide_numbers()
```
