```python
MAX_LENGTH = 3

def shorten(lst):
   while len(lst) > MAX_LENGTH :
     last_elem = lst.pop()
   print("Last eLement of list: " ,last_elem)

def get_list():
   lst = []
   elem = input("Enter an element of list: ")
   while elem != "" :
    lst.append(elem)
    elem = input("Enter an element of list: ")
   return lst

def main():
  lst = get_list()
  shorten(lst)

main()
```
